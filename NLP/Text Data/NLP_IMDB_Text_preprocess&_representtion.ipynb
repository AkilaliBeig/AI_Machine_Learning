```python
def expand_contractions(text):
    if contractions:
        return contractions.fix(text)
    else:
        return text


df['review'] = df['review'].apply(expand_contractions)
```

Chat / slang replacement

```python
chat_map = {
    "u": "you",
    "ur": "your",
    "btw": "by the way",
    "omg": "oh my god",
    "idk": "i do not know",
    "thx": "thanks",
    "pls": "please",
    "gr8": "great",
}


# Function to replace chat words
def chat_word_treatment(text):
  if not isinstance(text, str):   # if text is None or not string
        return ""                   # return empty string
  words = text.split()
  new_words = []
  for word in words:
    # replace if word exists in map, else keep as is
    new_words.append(chat_map.get(word.lower(), word))
  return " ".join(new_words)



# Apply on overview column
df['review'] = df['review'].apply(chat_word_treatment)
```

Remove URLs, punctuation & special character

```python
# remove punctuation (keeping whitespace)
table = str.maketrans('', '', string.punctuation)
df['review'] = df['review'].apply(lambda x: x.translate(table))
# remove extra non-alphanumeric chars (if needed)
df['review'] = df['review'].apply(lambda x: re.sub(r'[^0-9a-zA-Z\s]', ' ', x))
# collapse multiple spaces
df['review'] = df['review'].apply(lambda x: re.sub(r'\s+', ' ', x).strip())
```

Remove stopwords (but keep negations)

```python
from spacy.lang.en.stop_words import STOP_WORDS
import spacy

# Load spacy English model
nlp = spacy.load("en_core_web_sm")

# 1. Load SpaCy’s default stopwords
stop_words = spacy.lang.en.stop_words.STOP_WORDS

# 2. Keep important negations (because they change sentiment)
for w in ["no", "not", "never", "none", "n't"]:
    stop_words.discard(w)

# 3. Function to remove stopwords
def remove_stop_spacy(doc_text):
    doc = nlp(doc_text)  # tokenize using SpaCy
    tokens = [token.text for token in doc
              if token.text.lower() not in stop_words and not token.is_space]
    return " ".join(tokens)

# 4. Apply function to reviews column
df['review'] = df['review'].apply(remove_stop_spacy)
```

```python
df['review'][3]
```

✅ Step-by-step

Stopwords → Words like the, is, at, on, of, for that don’t carry much meaning in NLP are removed.
Example: "The movie is very good" → "movie good"

Why discard(w)?
Normally, SpaCy includes words like not, no, never in stopwords.
But for sentiment analysis (your IMDB dataset), these words are important!

"I like this movie" vs "I do **not** like this movie" → sentiment changes completely.
That’s why we keep negations.

remove_stop_spacy() function:

nlp(doc_text) → tokenizes text using SpaCy.

Removes stopwords (if token.text.lower() not in stop_words).

Removes spaces.

Joins back into cleaned text.

Apply on DataFrame → Every review is processed, and stopwords (except negations) are removed.

Lemmatization

```python
    def lemmatize_spacy(text):
        doc = nlp(text)
        lemmas = []
        for token in doc:
            if token.lemma_ == "-PRON-":       # spaCy uses -PRON- for pronouns
                lemmas.append(token.text.lower())
            else:
                lemmas.append(token.lemma_.lower())
        return " ".join(lemmas)
    df['clean_review'] = df['review'].apply(lemmatize_spacy)
```

Bag of Words (BoW)

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder
```

```python
# Use cleaned text column
texts = df['clean_review'].astype(str).tolist()

# Build CountVectorizer (uni-grams)
cv = CountVectorizer(ngram_range=(1,1), min_df=1)  # min_df can be increased to remove rare words
X_counts = cv.fit_transform(texts)  # sparse matrix: (n_docs, n_features)
```

```python
X_counts
```

```python
# Vocabulary (word -> column index)
vocab = cv.vocabulary_        # dict: word -> index
vocab_size = len(vocab)
print("Vocabulary size (uni-gram):", vocab_size)
```

```python
# Total occurrences of each word across the whole corpus
word_counts = np.asarray(X_counts.sum(axis=0)).ravel()  # array aligned with cv.get_feature_names_out()
features = cv.get_feature_names_out()
counts_df = pd.DataFrame({'word': features, 'count': word_counts})
counts_df = counts_df.sort_values('count', ascending=False).reset_index(drop=True)

# Show top 20 words
print(counts_df.head(20))
```

Bag of Bi-gram and Tri-gram (and dimensionality observation)

```python
# Bi-grams
cv_bigram = CountVectorizer(ngram_range=(2,2), min_df=1)
X_bigram = cv_bigram.fit_transform(texts)
vocab_bigram = cv_bigram.get_feature_names_out()
print("Bi-gram vocab size:", len(vocab_bigram))

# Tri-grams
cv_trigram = CountVectorizer(ngram_range=(3,3), min_df=1)
X_trigram = cv_trigram.fit_transform(texts)
vocab_trigram = cv_trigram.get_feature_names_out()
print("Tri-gram vocab size:", len(vocab_trigram))

# Show top 20 bi-grams and their counts
counts_bigram = np.asarray(X_bigram.sum(axis=0)).ravel()
bigram_df = pd.DataFrame({'ngram': vocab_bigram, 'count': counts_bigram}).sort_values('count', ascending=False)
print(bigram_df.head(20))
```

```python

# Show top 20 bi-grams and their counts
counts_trigram = np.asarray(X_trigram.sum(axis=0)).ravel()
trigram_df = pd.DataFrame({'ngram': vocab_trigram, 'count': counts_trigram}).sort_values('count', ascending=False)
print(trigram_df.head(20))
```

```python
print(counts_df.count())
print(bigram_df.count())
print(trigram_df.count())
```

TF–IDF

```python
tfidf = TfidfVectorizer(ngram_range=(1,1), min_df=1)  # uni-gram TF-IDF
X_tfidf = tfidf.fit_transform(texts)

# Vocabulary and idf values
features = tfidf.get_feature_names_out()
idf_values = tfidf.idf_   # array aligned with features

idf_df = pd.DataFrame({'word': features, 'idf': idf_values}).sort_values('idf', ascending=False)
print("Vocabulary size (tfidf uni-gram):", len(features))
print(idf_df.head(20))  # top 20 by idf (rare words)
```
